# Graph Basics

    A graph consists of vertices (nodes) and edges connecting pairs of vertices.

    ## How the Algorithm / Data Structure Works


Graphs may be:

- directed or undirected
- weighted or unweighted

They are useful for representing networks, relationships, routes, and
dependencies.



````python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A"],
    "D": ["B"],
}

print(graph)
````

## Output

    ```text
    {'A': ['B', 'C'], 'B': ['A', 'D'], 'C': ['A'], 'D': ['B']}
    ```

    ## Complexity Analysis


With an adjacency list, storing the graph takes **O(V + E)** space for
a typical sparse graph.
