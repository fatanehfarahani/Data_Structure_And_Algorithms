# Depth-First Search (DFS)

    DFS explores as far as possible along one branch before backtracking.

    ## How the Algorithm / Data Structure Works


DFS can be implemented recursively or with an explicit stack.

The iterative version below makes the stack visible.



````python
def dfs(graph, start):
    visited = set()
    order = []
    stack = [start]

    while stack:
        node = stack.pop()

        if node in visited:
            continue

        visited.add(node)
        order.append(node)

        for neighbor in reversed(graph.get(node, [])):
            if neighbor not in visited:
                stack.append(neighbor)

    return order


graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E"],
    "D": [],
    "E": [],
}

print(dfs(graph, "A"))
````

## Output

    ```text
    ['A', 'B', 'D', 'C', 'E']
    ```

    ## Complexity Analysis


**Time Complexity: O(V + E)**

**Space Complexity: O(V)**

DFS is useful for connectivity, cycle-related problems, and many
backtracking-style graph problems.
