# Adjacency List

    An adjacency list stores, for each vertex, a collection of its neighboring vertices.

    ## How the Algorithm / Data Structure Works


1. Create one list for each vertex.
2. Add an edge by adding the destination to the appropriate neighbor list.
3. For an undirected graph, add both directions.



````python
graph = {}

def add_edge(graph, a, b):
    graph.setdefault(a, []).append(b)
    graph.setdefault(b, []).append(a)


add_edge(graph, "A", "B")
add_edge(graph, "A", "C")

print(graph)
````

## Output

    ```text
    {'A': ['B', 'C'], 'B': ['A'], 'C': ['A']}
    ```

    ## Complexity Analysis


Building an adjacency list is typically **O(V + E)**.

Neighbor iteration for a vertex takes time proportional to its degree.
