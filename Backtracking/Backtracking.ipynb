
Backtracking

# Backtracking explores possible choices and undoes a choice when the current partial solution cannot lead to a ** valid solution **.
The general pattern is:

Choose a candidate.
Add it to the current state.
Explore recursively.
Undo the choice.
This is especially useful for constraint and combinatorial problems.


```python
def permutations(values):
    result = []
    path = []
    used = [False] * len(values)

    def backtrack():
        if len(path) == len(values):
            result.append(path.copy())
            return

        for i, value in enumerate(values):
            if used[i]:
                continue

            used[i] = True
            path.append(value)

            backtrack()

            path.pop()
            used[i] = False

    backtrack()
    return result


print(permutations([1, 2, 3]))
```

The recursion depth is **O(n)**, excluding the memory needed to store the complete result.
