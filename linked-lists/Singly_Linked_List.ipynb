# Singly Linked List

    A singly linked list consists of nodes. Each node stores a value and a reference to the next node.

    ## How the Algorithm / Data Structure Works


1. The list has a `head`.
2. Each node points to the next node.
3. The final node points to `None`.
4. Traversal follows the references one by one.



````python
class Node:
    def __init__(self, value, next_node=None):
        self.value = value
        self.next = next_node


head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

current = head
while current is not None:
    print(current.value)
    current = current.next
````

## Output

        ```text
        10
20
30
        ```

        ## Complexity Analysis


    **Access/Search: O(n)**

    **Insert at head: O(1)**

    **Traversal: O(n)**

    The list does not provide constant-time indexed access like an array.
