# Reverse a Singly Linked List

    Reversing a linked list changes the direction of each `next` reference.

    ## How the Algorithm / Data Structure Works


1. Keep a `previous` pointer.
2. Save the current node's next pointer.
3. Reverse the current pointer.
4. Move forward.
5. Continue until the end.



````python
class Node:
    def __init__(self, value, next_node=None):
        self.value = value
        self.next = next_node


def reverse(head):
    previous = None
    current = head

    while current is not None:
        following = current.next
        current.next = previous
        previous = current
        current = following

    return previous


head = Node(1, Node(2, Node(3)))

head = reverse(head)

current = head
while current:
    print(current.value)
    current = current.next
````

## Output

        ```text
        3
2
1
        ```

        ## Complexity Analysis


    **Time Complexity: O(n)**

    **Auxiliary Space Complexity: O(1)**
