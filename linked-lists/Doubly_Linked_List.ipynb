# Doubly Linked List

    A doubly linked list stores references to both the next and previous nodes.

    ## How the Algorithm / Data Structure Works


Each node contains:

`previous ← node → next`

The additional reference allows traversal in both directions.



````python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None


first = Node(10)
second = Node(20)

first.next = second
second.prev = first

print(first.value, "->", first.next.value)
print(second.value, "<-", second.prev.value)
````

## Output

        ```text
        10 -> 20
20 <- 10
        ```

        ## Complexity Analysis


    With a reference to the relevant node, insertion/removal can be **O(1)**.
    Searching for a value remains **O(n)**.

    The extra `prev` reference increases memory usage compared with a singly
    linked list.
