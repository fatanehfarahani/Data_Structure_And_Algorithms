# Divide and Conquer

    Divide and conquer solves a problem by dividing it into smaller subproblems, solving them, and combining their results.

    ## How the Algorithm / Data Structure Works


The general pattern is:

1. Divide.
2. Solve recursively.
3. Combine.

Merge Sort is a classic example.



````python
def sum_divide_and_conquer(values):
    if not values:
        return 0

    if len(values) == 1:
        return values[0]

    middle = len(values) // 2
    left = sum_divide_and_conquer(values[:middle])
    right = sum_divide_and_conquer(values[middle:])

    return left + right


print(sum_divide_and_conquer([1, 2, 3, 4, 5]))
````

## Output

    ```text
    15
    ```

    ## Complexity Analysis


The recurrence here leads to **O(n)** time because every element is
eventually included once.

The example is intended to illustrate the structure rather than provide
the most memory-efficient implementation.
