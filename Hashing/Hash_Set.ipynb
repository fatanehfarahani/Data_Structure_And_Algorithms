# Hash Set

    A set stores unique elements and is useful for fast membership testing.

    ## How the Algorithm / Data Structure Works


1. Add values to the set.
2. Duplicate values are ignored.
3. Membership can be checked efficiently.



````python
values = {2, 4, 6, 6, 8}

print(values)
print(4 in values)
print(5 in values)
````

## Output

        ```text
        {2, 4, 6, 8}
True
False
        ```

        ## Complexity Analysis


    Average-case membership, insertion, and deletion are **O(1)**.

    Space usage is **O(n)** for n stored elements.
