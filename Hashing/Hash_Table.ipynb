# Hash Table

    A hash table maps keys to values using a hash function. Python's `dict` is a hash-table-based mapping.

    ## How the Algorithm / Data Structure Works


1. A key is transformed into a hash value.
2. The hash determines where the entry is stored.
3. Lookup uses the key to locate the value.

Hash tables are especially useful when fast average-case lookup is needed.



````python
scores = {
    "Alice": 92,
    "Bob": 85,
    "Carol": 97,
}

print(scores["Bob"])

scores["David"] = 88
print(scores)
````

## Output

        ```text
        85
{'Alice': 92, 'Bob': 85, 'Carol': 97, 'David': 88}
        ```

        ## Complexity Analysis


    Typical average-case:

    - Lookup: **O(1)**
    - Insert: **O(1)**
    - Delete: **O(1)**

    These are average-case expectations; worst-case behavior depends on the
    hashing implementation and collision behavior.
