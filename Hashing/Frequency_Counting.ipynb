# Frequency Counting

    Frequency counting uses a hash map to count how often each value occurs.

    ## How the Algorithm / Data Structure Works


1. Start with an empty dictionary.
2. Read each value.
3. Increment its count.
4. Return the completed frequency table.



````python
def frequency_count(values):
    counts = {}

    for value in values:
        counts[value] = counts.get(value, 0) + 1

    return counts


data = ["a", "b", "a", "c", "b", "a"]
print(frequency_count(data))
````

## Output

    ```text
    {'a': 3, 'b': 2, 'c': 1}
    ```

    ## Complexity Analysis


**Time Complexity: O(n)** average-case.

**Space Complexity: O(k)** where `k` is the number of distinct values.
