# Heap Sort

    Heap Sort uses a heap structure to repeatedly select the next largest or smallest element.

    ## How the Algorithm / Data Structure Works


A common approach is:

1. Build a max heap.
2. Move the maximum element to the end.
3. Restore the heap property.
4. Repeat for the remaining elements.



````python
def heap_sort(values):
    values = list(values)
    n = len(values)

    def sift_down(root, size):
        while True:
            child = 2 * root + 1

            if child >= size:
                return

            if child + 1 < size and values[child] < values[child + 1]:
                child += 1

            if values[root] >= values[child]:
                return

            values[root], values[child] = values[child], values[root]
            root = child

    for start in range(n // 2 - 1, -1, -1):
        sift_down(start, n)

    for end in range(n - 1, 0, -1):
        values[0], values[end] = values[end], values[0]
        sift_down(0, end)

    return values


data = [4, -1, 9, 2, 7, 3]
print(heap_sort(data))
````

## Output

    ```text
    [-1, 2, 3, 4, 7, 9]
    ```

    ## Complexity Analysis


**Best Case: O(n log n)**

**Average Case: O(n log n)**

**Worst Case: O(n log n)**

**Space Complexity: O(1)** auxiliary space for the iterative in-place
implementation.

Heap Sort is not stable in its standard in-place form.
