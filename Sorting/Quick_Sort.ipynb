# Quick Sort

    Quick Sort is a divide-and-conquer sorting algorithm that partitions an array around a pivot.

    ## How the Algorithm / Data Structure Works


1. Choose a pivot.
2. Partition the array around the pivot.
3. Recursively sort the left partition.
4. Recursively sort the right partition.



````python
def quick_sort(values):
    values = list(values)

    def partition(left, right):
        pivot = values[right]
        boundary = left

        for index in range(left, right):
            if values[index] <= pivot:
                values[index], values[boundary] = values[boundary], values[index]
                boundary += 1

        values[boundary], values[right] = values[right], values[boundary]
        return boundary

    def sort(left, right):
        if left >= right:
            return

        pivot_index = partition(left, right)
        sort(left, pivot_index - 1)
        sort(pivot_index + 1, right)

    sort(0, len(values) - 1)
    return values


data = [7, 2, 9, 1, 5, 3]
print(quick_sort(data))
````

## Output

    ```text
    [1, 2, 3, 5, 7, 9]
    ```

    ## Complexity Analysis


**Best Case: O(n log n)**

**Average Case: O(n log n)**

**Worst Case: O(n²)** when partitions are consistently very unbalanced.

**Average auxiliary space: O(log n)** for the recursion stack.

The exact behavior depends strongly on pivot selection.
