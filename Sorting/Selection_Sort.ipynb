# Selection Sort

    Selection Sort repeatedly selects the smallest remaining element and places it at the next position.

    ## How the Algorithm / Data Structure Works


1. Consider the first unsorted position.
2. Find the smallest value in the remaining part.
3. Swap it with the value at the unsorted position.
4. Move the boundary of the sorted region.
5. Repeat.



````python
def selection_sort(values):
    values = list(values)

    n = len(values)

    for i in range(n):
        smallest = i

        for j in range(i + 1, n):
            if values[j] < values[smallest]:
                smallest = j

        values[i], values[smallest] = values[smallest], values[i]

    return values


data = [4, -1, 5, 2, 0]
print(selection_sort(data))
````

## Output

    ```text
    [-1, 0, 2, 4, 5]
    ```

    ## Complexity Analysis


**Best Case: O(n²)**

**Average Case: O(n²)**

**Worst Case: O(n²)**

**Space Complexity: O(1)** auxiliary space.

Selection Sort performs few swaps but still makes quadratic comparisons.
