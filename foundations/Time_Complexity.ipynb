# Time Complexity

    Time complexity is an estimate of how the number of operations grows with input size.

    ## How the Algorithm / Data Structure Works


1. Determine what the algorithm loops over.
2. Count nested loops.
3. Consider whether the input is divided or reduced.
4. Express the dominant growth using Big-O notation.



````python
def find_max(values):
    maximum = values[0]

    for value in values[1:]:
        if value > maximum:
            maximum = value

    return maximum


data = [7, 2, 9, 4, 1]
print(find_max(data))
````

## Output

    ```text
    9
    ```

    ## Complexity Analysis


The algorithm examines each element once.

**Time Complexity: O(n)**

**Space Complexity: O(1)** auxiliary space.
