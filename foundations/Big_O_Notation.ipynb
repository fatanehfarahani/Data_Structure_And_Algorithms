# Big-O Notation

    Big-O notation describes how the running time or memory requirement of an algorithm grows as the input size increases.

    ## How the Algorithm / Data Structure Works


1. Identify the input size, usually represented by `n`.
2. Identify the operation that dominates the work.
3. Ignore constant factors for asymptotic analysis.
4. Keep the fastest-growing term.

Common growth rates are:

`O(1)`, `O(log n)`, `O(n)`, `O(n log n)`, and `O(n²)`.



````python
def constant_time(values):
    return values[0]


def linear_time(values):
    total = 0
    for value in values:
        total += value
    return total


def quadratic_time(values):
    count = 0
    for _ in values:
        for _ in values:
            count += 1
    return count


data = [1, 2, 3, 4]

print("O(1):", constant_time(data))
print("O(n):", linear_time(data))
print("O(n^2):", quadratic_time(data))
````

## Output

        ```text
        O(1): 1
O(n): 10
O(n^2): 16
        ```

        ## Complexity Analysis


    - `constant_time` → **O(1)**
    - `linear_time` → **O(n)**
    - `quadratic_time` → **O(n²)**

    The examples are intentionally small so that the growth pattern is easy
    to inspect.
