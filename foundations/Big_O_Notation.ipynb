Big-O Notation
**Big-O** notation describes how the running time or memory requirement of an algorithm grows as the input size increases.

How the Algorithm / Data Structure Works
Identify the input size, usually represented by n.
Identify the operation that dominates the work.
Ignore constant factors for asymptotic analysis.
Keep the fastest-growing term.
Common growth rates are:

**O(1), O(log n), O(n), O(n log n), and O(n²)**.

```python
def constant_time(values):
    return values[0]


def linear_time(values):
    total = 0
    for value in values:
        total += value
    return total


def quadratic_time(values):
    count = 0
    for _ in values:
        for _ in values:
            count += 1
    return count


data = [1, 2, 3, 4]

print("O(1):", constant_time(data))
print("O(n):", linear_time(data))
print("O(n^2):", quadratic_time(data))
```

-constant_time → **O(1)**
-linear_time → **O(n)**
-quadratic_time → **O(n²)**
