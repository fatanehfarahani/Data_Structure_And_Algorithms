# Recursion Basics

    Recursion solves a problem by reducing it to smaller versions of the same problem.

    ## How the Algorithm / Data Structure Works


A recursive function normally contains:

1. A **base case**.
2. A **recursive case**.
3. Progress toward the base case.

In the example below, `n` decreases until it reaches 1.



````python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n <= 1:
        return 1

    return n * factorial(n - 1)


print(factorial(5))
````

## Output

    ```text
    120
    ```

    ## Complexity Analysis


There are `n` recursive calls in the general case.

**Time Complexity: O(n)**

**Auxiliary Space Complexity: O(n)** because of the recursion stack.
