# Frequency Map Pattern

    A frequency map records how often values occur and is useful for many counting and matching problems.

    ## How the Algorithm / Data Structure Works


1. Create an empty map.
2. Traverse the input.
3. Increment the count for each value.
4. Use the resulting map for later decisions.



````python
def frequency_map(values):
    counts = {}

    for value in values:
        counts[value] = counts.get(value, 0) + 1

    return counts


print(frequency_map([1, 2, 1, 3, 2, 1]))
````

## Output

    ```text
    {1: 3, 2: 2, 3: 1}
    ```

    ## Complexity Analysis


**Time Complexity: O(n)** average-case.

**Space Complexity: O(k)** where `k` is the number of distinct values.
