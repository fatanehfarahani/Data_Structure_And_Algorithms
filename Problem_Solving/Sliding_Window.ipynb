# Sliding Window

    The sliding-window pattern maintains a moving contiguous region of an array or string.

    ## How the Algorithm / Data Structure Works


For a fixed-size window:

1. Calculate the first window.
2. Move the window one position.
3. Remove the value leaving the window.
4. Add the value entering the window.



````python
def max_sum_fixed_window(values, k):
    window = sum(values[:k])
    best = window

    for right in range(k, len(values)):
        window += values[right] - values[right - k]
        best = max(best, window)

    return best


print(max_sum_fixed_window([2, 1, 5, 1, 3, 2], 3))
````

## Output

    ```text
    9
    ```

    ## Complexity Analysis


Each element enters and leaves the window at most once.

**Time Complexity: O(n)**

**Space Complexity: O(1)**
