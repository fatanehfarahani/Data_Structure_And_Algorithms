# Prefix Sum

    A prefix-sum array stores cumulative totals so that range sums can be calculated quickly.

    ## How the Algorithm / Data Structure Works


If:

`prefix[i] = sum(values[:i])`

then a range sum can be obtained using subtraction of two prefix values.



````python
def build_prefix_sum(values):
    prefix = [0]

    for value in values:
        prefix.append(prefix[-1] + value)

    return prefix


values = [2, 4, 1, 5, 3]
prefix = build_prefix_sum(values)

left = 1
right = 4

range_sum = prefix[right + 1] - prefix[left]

print(prefix)
print(range_sum)
````

## Output

        ```text
        [0, 2, 6, 7, 12, 15]
10
        ```

        ## Complexity Analysis


    Building the prefix array takes **O(n)** time and **O(n)** extra space.

    Each subsequent range-sum query takes **O(1)** time.
