# Two Pointers Pattern

    The two-pointer pattern maintains two indices and moves them according to the structure of the problem.

    ## How the Algorithm / Data Structure Works


A common example uses a sorted array:

1. Put one pointer at the beginning.
2. Put one pointer at the end.
3. Compare the sum with the target.
4. Move the left or right pointer accordingly.



````python
def pair_sum_sorted(values, target):
    left = 0
    right = len(values) - 1

    while left < right:
        total = values[left] + values[right]

        if total == target:
            return left, right

        if total < target:
            left += 1
        else:
            right -= 1

    return None


print(pair_sum_sorted([1, 2, 4, 6, 8, 9], 10))
````

## Output

    ```text
    (1, 5)
    ```

    ## Complexity Analysis


**Time Complexity: O(n)**

**Space Complexity: O(1)**

The sorted-order assumption is important for the pointer movement rule.
