# Recursive Factorial

    The factorial function is a simple example of a problem whose definition naturally leads to recursion.

    ## How the Algorithm / Data Structure Works


The mathematical definition is:

`n! = n × (n-1)!`

with the base case:

`0! = 1`



````python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n == 0:
        return 1

    return n * factorial(n - 1)


print(factorial(5))
````

## Output

    ```text
    120
    ```

    ## Complexity Analysis


**Time Complexity: O(n)**

**Space Complexity: O(n)** due to recursive calls.
