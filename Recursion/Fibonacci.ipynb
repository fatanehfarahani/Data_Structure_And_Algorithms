# Recursive Fibonacci

    The Fibonacci sequence demonstrates both the simplicity and the cost of naive recursion.

    ## How the Algorithm / Data Structure Works


The recursive definition is:

`F(n) = F(n-1) + F(n-2)`

with `F(0)=0` and `F(1)=1`.



````python
def fibonacci(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(8))
````

## Output

    ```text
    21
    ```

    ## Complexity Analysis


The naive recursive implementation has exponential time complexity,
commonly described as **O(2^n)**.

Its recursion depth is **O(n)**.

This motivates memoization and dynamic programming.
