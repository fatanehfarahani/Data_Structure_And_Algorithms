# Strings

    A string is a sequence of characters. Many algorithmic problems reduce to scanning, counting, comparing, or transforming strings.

    ## How the Algorithm / Data Structure Works


1. Traverse the characters.
2. Track the information needed by the problem.
3. Build or compare the result.



````python
text = "algorithm"

print("Length:", len(text))
print("First character:", text[0])
print("Reversed:", text[::-1])
````

## Output

        ```text
        Length: 9
First character: a
Reversed: mhtirogla
        ```

        ## Complexity Analysis


    Indexing is typically **O(1)**.

    Creating a reversed string requires **O(n)** time and additional space.
