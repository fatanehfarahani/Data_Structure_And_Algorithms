# Arrays

    An array stores elements in an indexed sequence. Python's `list` is a dynamic-array implementation with fast indexed access.

    ## How the Algorithm / Data Structure Works


1. Elements are stored in a logical sequence.
2. An index identifies a position.
3. Access by index is typically constant time.
4. Inserting into the middle may require shifting later elements.



````python
values = [10, 20, 30, 40, 50]

print("First element:", values[0])
print("Third element:", values[2])

values.append(60)
print("After append:", values)
````

## Output

        ```text
        First element: 10
Third element: 30
After append: [10, 20, 30, 40, 50, 60]
        ```

        ## Complexity Analysis


    **Indexed access: O(1)**

    **Append: O(1) amortized**

    **Insertion/deletion in the middle: O(n)** in the general case.
