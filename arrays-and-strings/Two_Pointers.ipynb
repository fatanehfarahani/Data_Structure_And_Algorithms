# Two Pointers

    The two-pointer technique uses two indices to process a sequence without repeatedly scanning the same elements.

    ## How the Algorithm / Data Structure Works


A common example is checking whether a sequence is a palindrome:

1. Put one pointer at the beginning.
2. Put the other at the end.
3. Compare the two values.
4. Move both pointers toward the center.



````python
def is_palindrome(text):
    left = 0
    right = len(text) - 1

    while left < right:
        if text[left] != text[right]:
            return False

        left += 1
        right -= 1

    return True


print(is_palindrome("level"))
print(is_palindrome("algorithm"))
````

## Output

        ```text
        True
False
        ```

        ## Complexity Analysis


    Each character is examined at most once.

    **Time Complexity: O(n)**

    **Space Complexity: O(1)** auxiliary space.
