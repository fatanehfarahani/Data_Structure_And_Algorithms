# Common Array Operations

    Basic array operations are important because many later algorithms depend on them.

    ## How the Algorithm / Data Structure Works


Common operations include:

1. Access
2. Traversal
3. Search
4. Append
5. Insert
6. Delete



````python
values = [3, 7, 2, 9]

values.append(5)
print("Append:", values)

values.insert(1, 10)
print("Insert:", values)

values.remove(2)
print("Delete:", values)

print("Search index:", values.index(9))
````

## Output

        ```text
        Append: [3, 7, 2, 9, 5]
Insert: [3, 10, 7, 2, 9, 5]
Delete: [3, 10, 7, 9, 5]
Search index: 3
        ```

        ## Complexity Analysis


    Typical complexities:

    - Access by index: **O(1)**
    - Linear search: **O(n)**
    - Append: **O(1) amortized**
    - Middle insertion/deletion: **O(n)**
