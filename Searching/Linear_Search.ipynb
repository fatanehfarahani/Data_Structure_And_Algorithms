# Linear Search

    Linear Search checks elements one by one until the target is found or the sequence ends.

    ## How the Algorithm / Data Structure Works


1. Start at the first element.
2. Compare it with the target.
3. If it matches, return its position.
4. Otherwise continue.



````python
def linear_search(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index

    return -1


data = [8, 3, 10, 4, 7]

print(linear_search(data, 4))
print(linear_search(data, 9))
````

## Output

        ```text
        3
-1
        ```

        ## Complexity Analysis


    **Best Case: O(1)**

    **Average Case: O(n)**

    **Worst Case: O(n)**

    **Space Complexity: O(1)**
