# Binary Search

    Binary Search repeatedly halves a sorted search interval.

    ## How the Algorithm / Data Structure Works


1. Set left and right boundaries.
2. Calculate the middle index.
3. If the middle value is the target, stop.
4. If the middle value is too small, search the right half.
5. Otherwise search the left half.



````python
def binary_search(values, target):
    left = 0
    right = len(values) - 1

    while left <= right:
        middle = (left + right) // 2

        if values[middle] == target:
            return middle

        if values[middle] < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1


data = [1, 3, 5, 7, 9, 11]

print(binary_search(data, 7))
print(binary_search(data, 8))
````

## Output

        ```text
        3
-1
        ```

        ## Complexity Analysis


    The input must be sorted.

    **Best Case: O(1)**

    **Average Case: O(log n)**

    **Worst Case: O(log n)**

    **Space Complexity: O(1)** for the iterative version.
