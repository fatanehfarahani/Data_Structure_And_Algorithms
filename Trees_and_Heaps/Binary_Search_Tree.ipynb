# Binary Search Tree

    A Binary Search Tree (BST) organizes values so that smaller values are placed to the left and larger values to the right.

    ## How the Algorithm / Data Structure Works


For each node:

- left subtree values are smaller
- right subtree values are larger

This ordering allows search to eliminate part of the tree at each step
when the tree is reasonably balanced.



````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


def insert(root, value):
    if root is None:
        return Node(value)

    if value < root.value:
        root.left = insert(root.left, value)
    else:
        root.right = insert(root.right, value)

    return root


def search(root, target):
    while root is not None:
        if root.value == target:
            return True
        root = root.left if target < root.value else root.right

    return False


root = None
for value in [8, 3, 10, 1, 6]:
    root = insert(root, value)

print(search(root, 6))
print(search(root, 9))
````

## Output

        ```text
        True
False
        ```

        ## Complexity Analysis


    **Average search/insert: O(log n)** for a balanced tree.

    **Worst case: O(n)** when the tree becomes highly unbalanced.
