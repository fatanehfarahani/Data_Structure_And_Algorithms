# Heap

    A heap is a complete binary-tree-based structure commonly used to implement priority queues.

    ## How the Algorithm / Data Structure Works


In a min-heap, the smallest value is at the root.

Python's `heapq` module provides a min-heap.



````python
import heapq

values = [7, 2, 9, 1, 5]

heapq.heapify(values)

print("Heap:", values)
print("Smallest:", heapq.heappop(values))
print("Remaining:", values)
````

## Output

        ```text
        Heap: [1, 2, 9, 7, 5]
Smallest: 1
Remaining: [2, 5, 9, 7]
        ```

        ## Complexity Analysis


    - Peek minimum: **O(1)**
    - Insert: **O(log n)**
    - Remove minimum: **O(log n)**
    - Heapify: **O(n)**
