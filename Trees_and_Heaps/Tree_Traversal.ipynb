# Tree Traversal

    Tree traversal means visiting the nodes of a tree in a systematic order.

    ## How the Algorithm / Data Structure Works


The three classic depth-first traversals are:

- Preorder: Root → Left → Right
- Inorder: Left → Root → Right
- Postorder: Left → Right → Root

Inorder traversal of a BST visits values in sorted order.



````python
class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right


root = Node(
    4,
    left=Node(2, Node(1), Node(3)),
    right=Node(6, Node(5), Node(7))
)


def inorder(node):
    if node is None:
        return

    inorder(node.left)
    print(node.value, end=" ")
    inorder(node.right)


inorder(root)
````

## Output

    ```text
    1 2 3 4 5 6 7
    ```

    ## Complexity Analysis


Every node is visited once.

**Time Complexity: O(n)**

Recursive auxiliary space is **O(h)** where `h` is the tree height.
