# Binary Tree

    A binary tree is a hierarchical structure in which each node has at most two children.

    ## How the Algorithm / Data Structure Works


Each node may have:

- a left child
- a right child

A tree has a root, internal nodes, and leaves.



````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


root = Node(10)
root.left = Node(5)
root.right = Node(15)

print(root.value)
print(root.left.value)
print(root.right.value)
````

## Output

        ```text
        10
5
15
        ```

        ## Complexity Analysis


    Access to a known node is **O(1)**.

    Searching a general binary tree is **O(n)** in the worst case.

    Storage is **O(n)**.
