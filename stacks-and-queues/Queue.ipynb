
Queue

## A queue is a First-In, First-Out (FIFO) data structure.
The first element inserted is the first element removed.

Typical operations:

enqueue
dequeue
front/peek


```python
from collections import deque
queue = deque()

queue.append("A")
queue.append("B")
queue.append("C")

print("Queue:", queue)
print("Dequeued:", queue.popleft())
print("Queue:", queue)
```

**Enqueue: O(1)**

**Dequeue: O(1)**

**Space: O(n)**
