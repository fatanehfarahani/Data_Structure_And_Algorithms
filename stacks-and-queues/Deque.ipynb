# Deque

    A deque, or double-ended queue, supports insertion and removal at both ends.

    ## How the Algorithm / Data Structure Works


A deque is useful when both the front and back of a sequence need efficient
operations.



````python
from collections import deque

items = deque([2, 3])

items.appendleft(1)
items.append(4)

print(items)

print(items.popleft())
print(items.pop())
print(items)
````

## Output

        ```text
        deque([1, 2, 3, 4])
1
4
deque([2, 3])
        ```

        ## Complexity Analysis


    Operations at either end are **O(1)** with `collections.deque`.
