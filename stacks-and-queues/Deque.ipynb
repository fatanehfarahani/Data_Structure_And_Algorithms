## A deque, or double-ended queue, supports insertion and removal at both ends.

A deque is useful when both the front and back of a sequence need efficient operations.

```python
from collections import deque

items = deque([2, 3])

items.appendleft(1)
items.append(4)

print(items)

print(items.popleft())
print(items.pop())
print(items)
```


Operations at either end are **O(1)** with `collections.deque`.
