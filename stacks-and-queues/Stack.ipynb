# Stack

    A stack is a Last-In, First-Out (LIFO) data structure.

    ## How the Algorithm / Data Structure Works


The two main operations are:

- `push`: add an item to the top.
- `pop`: remove the most recently added item.

A stack can be visualized as a pile of plates.



````python
stack = []

stack.append("A")
stack.append("B")
stack.append("C")

print("Stack:", stack)
print("Popped:", stack.pop())
print("Stack:", stack)
````

## Output

        ```text
        Stack: ['A', 'B', 'C']
Popped: C
Stack: ['A', 'B']
        ```

        ## Complexity Analysis


    With a Python list:

    **Push: O(1) amortized**

    **Pop: O(1)**

    **Space: O(n)**
