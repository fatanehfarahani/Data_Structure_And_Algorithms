# Greedy Algorithms

    A greedy algorithm repeatedly makes a locally optimal choice with the hope that the sequence of choices leads to a global optimum.

    ## How the Algorithm / Data Structure Works


A greedy approach normally consists of:

1. Identify the available choices.
2. Choose the best local option according to a rule.
3. Commit to that choice.
4. Continue.

The important part is proving that the local choice is safe for the
particular problem.



````python
def activity_selection(activities):
    # activities are (start, finish), sorted by finish time
    selected = []
    last_finish = float("-inf")

    for start, finish in activities:
        if start >= last_finish:
            selected.append((start, finish))
            last_finish = finish

    return selected


activities = [(1, 3), (2, 5), (4, 7), (6, 9), (8, 10)]
print(activity_selection(activities))
````

## Output

    ```text
    [(1, 3), (4, 7), (8, 10)]
    ```

    ## Complexity Analysis


With activities already sorted by finish time, the selection pass is
**O(n)**.

If sorting is required first, the overall complexity becomes
**O(n log n)**.
