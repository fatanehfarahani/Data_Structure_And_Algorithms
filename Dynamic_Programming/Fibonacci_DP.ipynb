# Fibonacci with Constant Extra Space

    The Fibonacci recurrence only needs the previous two values, so the full DP table is unnecessary.

    ## How the Algorithm / Data Structure Works


1. Keep the previous two values.
2. Compute the next value.
3. Shift the two variables.
4. Repeat until reaching `n`.



````python
def fibonacci(n):
    if n <= 1:
        return n

    previous, current = 0, 1

    for _ in range(2, n + 1):
        previous, current = current, previous + current

    return current


print(fibonacci(30))
````

## Output

    ```text
    832040
    ```

    ## Complexity Analysis


**Time Complexity: O(n)**

**Space Complexity: O(1)**
