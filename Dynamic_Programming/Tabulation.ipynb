# Tabulation

    Tabulation solves dynamic-programming states from the smallest cases upward and stores the results in a table.

    ## How the Algorithm / Data Structure Works


For Fibonacci:

1. Initialize the base cases.
2. Fill the table from left to right.
3. Use previously computed states to build the next state.



````python
def fibonacci(n):
    if n <= 1:
        return n

    dp = [0] * (n + 1)
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]


print(fibonacci(30))
````

## Output

    ```text
    832040
    ```

    ## Complexity Analysis


**Time Complexity: O(n)**

**Space Complexity: O(n)**

The table can be reduced to O(1) space for Fibonacci because only the last
two states are needed.
