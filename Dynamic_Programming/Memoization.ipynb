# Memoization

    Memoization stores results of previously solved subproblems so that repeated work is avoided.

    ## How the Algorithm / Data Structure Works


1. Define the recursive state.
2. Check whether the state has already been solved.
3. Return the stored result if available.
4. Otherwise compute it and store the result.



````python
def fibonacci(n, memo=None):
    if memo is None:
        memo = {}

    if n <= 1:
        return n

    if n in memo:
        return memo[n]

    memo[n] = fibonacci(n - 1, memo) + fibonacci(n - 2, memo)
    return memo[n]


print(fibonacci(30))
````

## Output

    ```text
    832040
    ```

    ## Complexity Analysis


There are O(n) distinct states for Fibonacci.

**Time Complexity: O(n)**

**Space Complexity: O(n)** for the memo and recursion stack.
